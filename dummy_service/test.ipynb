```python
class A:
    def __init__(self):
        self.test = 1
        self.not_test =2
    
class B(A):
    def __init__(self):
        super().__init__()

a = A()
b = B()
print(a.test)
print(b.test)
```

```python
class A:
    def __init__(self):
        self.test = 1
        self.not_test =2
    
class B():
    def __init__(self,a):
        self.a = a

a = A()
b = B(a)
print(a.test)
print(b.a.test)
        
```
